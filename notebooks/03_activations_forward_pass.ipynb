#### Step 1: Plot ReLU, sigmoid, and tanh over a range of inputs to see how each transforms values.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image, display
```

```python
def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)
```

```python
x = np.linspace(-5, 5, 200)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x, relu(x), label='ReLU')
plt.plot(x, sigmoid(x), label='Sigmoid')
plt.plot(x, tanh(x), label='Tanh')

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel('Input')
plt.ylabel('Output')
plt.title('Activation Functions')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

#### Step 2: For the Phase 3 project task, decide the correct output activation and loss function, and justify both.

The Phase 3 cardiac-risk project is a binary classification task because the
target `cardiac_risk` contains two classes: 0 and 1.

Therefore:

- **Output activation:** Sigmoid
- **Loss function:** Binary Cross-Entropy

Sigmoid is appropriate because it maps the output to a value between 0 and 1,
which can be interpreted as the predicted probability of the positive class
(`cardiac_risk = 1`).

Binary cross-entropy is appropriate because it measures the difference between
the predicted probability and the true binary target.

#### Step 3: By hand or in NumPy, compute one forward pass for a tiny 2-layer network on a sample input.

```python
X = np.array([0.6, -0.2])

W1 = np.array([
    [0.5, -0.3],
    [0.8,  0.2]
])

b1 = np.array([0.1, 0.0])

W2 = np.array([0.7, -0.5])
b2 = 0.1
```

```python
z1 = X @ W1 + b1
a1 = relu(z1)

print("Hidden weighted sum:", z1)
print("Hidden activation:", a1)
```

```python
display(Image(filename='../data/images/relu.png'))
```

```python
z2 = a1 @ W2 + b2
y_pred = sigmoid(z2)

print("Output weighted sum:", z2)
print("Prediction:", y_pred)
```

```python
display(Image(filename='../data/images/sigmoid.png'))
```

#### Step 4: Document the choices and the forward-pass result in Markdown.

A small neural network with two input features, two hidden neurons, and one
output neuron was used to demonstrate forward propagation.

The hidden layer used **ReLU**, while the output neuron used **sigmoid**
because the target task is binary classification.

For the sample input:

`X = [0.6, -0.2]`

the hidden layer produced:

`z1 = [0.24, -0.22]`

After applying ReLU:

`a1 = [0.24, 0.0]`

The output layer produced:

`z2 = 0.268`

After applying sigmoid, the final prediction was approximately:

`0.5666`

This means that, for this illustrative forward pass, the network assigned a
predicted probability of approximately **56.66%** to the positive class.

For the Phase 3 cardiac-risk classification task, the planned output
configuration is therefore:

- Hidden-layer activation: **ReLU**
- Output-layer activation: **Sigmoid**
- Loss function: **Binary Cross-Entropy**
